```python
import pandas as pd
df1=pd.read_excel("HR_1 (1).xlsx")
df2=pd.read_excel("HR_2 (1).xlsx")
merged_file = pd.merge(df1, df2,
                     left_on='EmployeeNumber',  
                     right_on='Employee ID',how='inner')      
merged_file.to_excel('HR_merged_file.xlsx', index=False)
print("Merge completed successfully!")
```
